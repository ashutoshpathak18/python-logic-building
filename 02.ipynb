```python
#find the element that appears more than n/2 times,
#  where n is the length of the list.
li = [2, 2, 1, 1, 1, 2, 2]
n = len(li)
dic = {}
for num in li:
    if num in dic:
        dic[num] += 1
    else:
        dic[num] = 1
for i in dic:
    if dic[i] > n/2:
        print(i)
    
```

```python
# Product of Array Except Self
li = [1, 2, 3, 4]
new = []
for i in range(len(li)):
    prod = 1
    for j in range(len(li)):
        if i != j:
         prod *= li[j]
    new.append(prod)
print(new)
```

```python
# Find the Leaders in an Array
li = [16, 17, 4, 3, 5, 2]
leader = []
for i in range(len(li)):
    flag = False
    for j in range(i+1,len(li)):
        if li[i] < li[j]:
            flag = True
    if not flag:
        leader.append(li[i])
print(leader)
```

```python
# Find the intersection of two arrays, including duplicates.
li1 = [1, 2, 2, 3, 4]
li2 = [2, 2, 4, 6]
interaction = []
for i in li1:
    for j in li2:
        if i == j:
            interaction.append(i)
            li2.remove(j)
            break
print(interaction)
```

```python
# find the length of the longest sequence of consecutive integers.
li = [9, 1, 4, 7, 3, 2, 6, 5]
largest = 0
for num in li:
    count = 1
    current = num
    while current+1 in li:
        current+=1
        count+=1
    if count > largest:
        largest = count
print(largest)
```
