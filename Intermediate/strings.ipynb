```python
# Count Character

num = input("Enter a string :")
count = 0

for i in num:
    count+=1
print(f"{num} has {count} character ")
 
```

```python
# Count vowels

num = input("Enter a string :")
count = 0

for i in num:
    if i in ('a','e','i','o','u','A','E','I','O','U'):
      count+=1   
print(f"{num} has {count} vowel ")
```

```python
# character frequency

text = input("Enter the word :")
count = {}

for ch in text:
    if ch in count:
        count[ch] +=1
    else:
        count[ch] = 1
print(count)
```

```python
# remove space from string

str = input("Enter the string :")
new_str = ""

for ch in str:
    if ch !=' ':
        new_str = new_str + ch
print(new_str)
```

```python
# print character which appeared more than once

str = input("Enter the string :")
dic = {}

for ch in str:
    if ch in dic:
        dic[ch]+=1
    else:
        dic[ch]=1

for i in dic:
    if dic[i] > 1:
        print(i)
# input = programming
```

```python
# Anagram checker

str1 = input("Enter first word :")
str2 = input("Enter second word :")
dic1 = {}
dic2 = {}
if len(str1) != len(str2):
    print("not anagram")
else:
    for ch in str1:
        if ch in dic1:
            dic1[ch] +=1
        else:
            dic1[ch] = 1
    for ch in str2:
        if ch in dic2:
            dic2[ch] +=1
        else:
            dic2[ch] = 1
    if dic1 == dic2:
        print("anagram")
    else:
        print("not anagram")

#input = silent, listen
```

```python
# find longest word
# input = python is fucking good
# output = fucking

sent = input("Enter the sentence :")
words = sent.split()
longest = ""
for word in words:
    if len(word) > len(longest):
        longest = word
print(longest)
```

```python
# word frequency

text = input("enter the text :")
words = text.split()
dic = {}
for word in words:
    if word in dic:
        dic[word] += 1
    else:
        dic[word] = 1
highest = 0
word = ""
for i in dic:
    if dic[i] > highest:
      highest = dic[i]
      word = i
print(word)
```
