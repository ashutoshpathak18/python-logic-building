```python
# Count Character

num = input("Enter a string :")
count = 0

for i in num:
    count+=1
print(f"{num} has {count} character ")
 
```

```python
# Count vowels

num = input("Enter a string :")
count = 0

for i in num:
    if i in ('a','e','i','o','u','A','E','I','O','U'):
      count+=1   
print(f"{num} has {count} vowel ")
```

```python
# reverse string

text = input("Enter the text :")
rev_text = ''

for ch in text:
    rev_text = ch + rev_text
print(rev_text)
```

```python
# pelindrom 

text = input("Enter the text :")
rev_text = ''

for ch in text:
    rev_text = ch + rev_text
if rev_text == text:
    print("Pelindrom")
else:
    print("not pelindrom")
```

```python
# character frequency

text = input("Enter the word :")
count = {}

for ch in text:
    if ch in count:
        count[ch] +=1
    else:
        count[ch] = 1
print(count)
```

```python
# remove space from string

str = input("Enter the string :")
new_str = ""

for ch in str:
    if ch !=' ':
        new_str = new_str + ch
print(new_str)
```
