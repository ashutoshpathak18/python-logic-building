```python
# Linear Search

li = [10, 25, 7, 42, 18]
target = 42
found = False
for i,j in enumerate(li):
    if j == target:
        print("element found at index :",i)
        found = True
        break
if not found:
    print("Element not found")
```

```python
#count how many times the target appears.
List = [5, 2, 5, 8, 5, 3]
target = 5
found = False
count = 0
for i in List:
    if i == target:
        count +=1
        found = True
if not found:
    print("target does not exist")
else:
    print(f"{target} appears {count} times")
```

```python
# two sum problem
li = [2, 7, 11, 15]
target = 9
found = False
for num in li:
    for i in li:
        if num != i:
            if num + i == target:
                print(num,i)
                found = True
    if found:
        break
```
