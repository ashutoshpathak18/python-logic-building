```python
# Linear Search

li = [10, 25, 7, 42, 18]
target = 42
found = False
for i,j in enumerate(li):
    if j == target:
        print("element found at index :",i)
        found = True
        break
if not found:
    print("Element not found")
```

```python
#count how many times the target appears.
List = [5, 2, 5, 8, 5, 3]
target = 5
found = False
count = 0
for i in List:
    if i == target:
        count +=1
        found = True
if not found:
    print("target does not exist")
else:
    print(f"{target} appears {count} times")
```

```python
# find the maximum difference between two elements
#  where the larger element comes after the smaller element.
li = [7, 1, 5, 3, 6, 4]
diff = 0
for i in range(len(li)):
    for j in range(i+1,len(li)):
        if li[j] - li[i] > diff:
            diff = li[j] - li[i]
for i in range(len(li)):
    for j in range(i+1,len(li)):
        if li[j] - li[i] == diff:
            print(f"The maximum difference is {diff} and the numbers are {li[j]} and {li[i]}")
```
