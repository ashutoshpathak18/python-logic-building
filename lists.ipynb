```python
# find largest number in a list 

li = [23,43,2,435,54,75,86]
largest = 0
for i in li:
    if i > largest:
        largest = i
print(largest)
```

```python
# find the smallest

li = [-5, -2, -10, -1]
smallest = li[0]
for i in li:
    if i < smallest:
        smallest = i

print(smallest)
```

```python
# find second largest 

li = [10, 25, 7, 42, 18]
largest = 0
for i in li:
    if i > largest:
        largest = i
li.remove(largest)
lar = 0
for i in li:
    if i > lar:
        lar = i
print(lar)
```

```python
# remove duplicates

li = [1, 2, 2, 3, 1, 4, 3,5]
new_li = []
for i in li:
    if i not in new_li:
        new_li.append(i)
print(new_li)
```

```python
# count frequency

li = [1, 2, 2, 3, 1, 4, 3,5]
new = {}
for i in li:
    if i in new:
        new[i] +=1
    else:
        new[i] = 1
print(new)
```

```python
# find common elements

li1 = [1, 2, 3,3, 4, 5]
li2 = [3, 4, 5, 6, 7]
li3 = []
for i in li1:
    if i in li2:
        li3.append(i)
li = []
for element in li3:
    if element not in li:
        li.append(element)
print(li)
```
