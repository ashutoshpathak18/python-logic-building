```python
# find largest number in a list 

li = [23,43,2,435,54,75,86]
largest = 0
for i in li:
    if i > largest:
        largest = i
print(largest)
```

```python
# find the smallest

li = [-5, -2, -10, -1]
smallest = li[0]
for i in li:
    if i < smallest:
        smallest = i

print(smallest)
```

```python
# find second largest 

li = [10, 25, 7, 42, 18]
largest = 0
for i in li:
    if i > largest:
        largest = i
li.remove(largest)
lar = 0
for i in li:
    if i > lar:
        lar = i
print(lar)
```

```python
# remove duplicates

li = [1, 2, 2, 3, 1, 4, 3,5]
new_li = []
for i in li:
    if i not in new_li:
        new_li.append(i)
print(new_li)
```

```python
# count frequency

li = [1, 2, 2, 3, 1, 4, 3,5]
new = {}
for i in li:
    if i in new:
        new[i] +=1
    else:
        new[i] = 1
print(new)
```

```python
# find common elements

li1 = [1, 2, 3,3, 4, 5]
li2 = [3, 4, 5, 6, 7]
li3 = []
for i in li1:
    if i in li2:
        li3.append(i)
li = []
for element in li3:
    if element not in li:
        li.append(element)
print(li)
```

```python
# find missing value

li = [2, 3, 1, 5]
large = 0
for num in li:
    if num > large:
        large = num
small = li[0]
for i in li:
    if i < small:
        small = i
for i in range(small,large+1):
    if i not in li:
        print("missing element is ",i)
```

```python
# move zeros to end

li =  [0, 0, 1, 2]
li1 = []
for i in li:
    if i != 0:
        li1.append(i)
for i in li:
    if i == 0:
        li1.append(i)
print(li1)
```

```python
# find first not repeating element

li =  [4, 5, 1,3, 2, 1, 4, 5]
dic = {}
for i in li:
    if i in dic:
        dic[i] += 1
    else:
        dic[i] = 1
for i in dic:
    if dic[i] == 1:
        print("first non repeating element :",i)
        break
```
