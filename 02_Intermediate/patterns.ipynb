```python
# Square pattern

n = int(input("Enter the number of rows :"))

for i in range(0,n):
        print("*"*n)
```

```python
# Increasing triangle pattern

n = int(input("Enter the number of rows of the triangle :"))

for i in range(1,n+1):
        print("*"*i)
```

```python
# Decreasing Triangle

n = int(input("Enter the number of rows of the triangle :"))

for i in range(n,0,-1):
    print("*"*i)
```

```python
# Number Triangle

n = int(input("Enter the number of rows of the triangle :"))

for i in range(1,n+1):
    print(f"{i}"*i)
```

```python
# Number Triangle

n = int(input("Enter the number of rows of the triangle :"))

for i in range(1,n+1):
    for j in range(1,i+1):
        print(j,end=' ')
    print()
```

```python
n =int(input("Enter the number of rows :"))
i = 1
while n>0:
    print(" "*(n-1),end=" ")
    print("*"*i)
    n-=1
    i+=1
```

```python
# using nested loop

n = int(input("enter the number of rows :"))
k = 1
for i in range(n,0,-1):
    for j in range(1,i):
        print(" ",end=" ")
    print("*"*k)
    k+=1
```
